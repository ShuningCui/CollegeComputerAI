## <font color = #0288d1>文件操作</font>
### 文件打开模式
### 在Python中，使用`open()`函数来打开文件，其基本语法为：
###    `file_object = open(file_path, mode)`
### 其中，`file_path` 是文件的路径，可以是绝对路径或相对路径；`mode`是文件的打开模式，常见的模式有：
- ### <font color="purple">**'r'：只读模式**</font>，默认值。用于读取文件内容，如果文件不存在会抛出  `FileNotFoundError` 异常。
- ### <font color="purple">**'w'：写入模式**</font>。如果文件存在，会清空文件内容；如果文件不存在，则创建新文件。
- ### <font color="purple">**'a'：追加模式**</font>。如果文件存在，会在文件末尾追加内容；如果文件不存在，则创建新文件。
- ### <font color="purple">**'x'：独占创建模式**</font>。如果文件已经存在，会抛出 `FileExistsError`异常；如果文件不存在，则创建新文件。
- ### <font color="purple">**'b'：二进制模式**</font>。可以与其他模式结合使用，如 `'rb'`表示以二进制只读模式打开文件。
- ### <font color="purple">**'t'：文本模式**</font>，默认值。可以与其他模式结合使用，如 `'wt'`表示以文本写入模式打开文件。
- ### <font color="purple">**'+'：读写模式**</font>。可以与其他模式结合使用，如 `'r+'`表示以读写模式打开文件。


### 文本文件的读写操作
### 可以采用<font color="purple">**逐行读取**</font>的方式将一个文本文件中的内容读取出来：

```python
try:
    with open('example.txt', 'r', encoding='utf-8') as file:
        for line in file:
            print(line.strip())  # 去除每行末尾的换行符
except FileNotFoundError:
    print("文件未找到！")
```

```python
# 读取整个文件内容为一个字符串：

try:
    with open('example.txt', 'r', encoding='utf-8') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("文件未找到！")
```

```python
# 读取所有行到一个列表中：

try:
    with open('example.txt', 'r', encoding='utf-8') as file:
        lines = file.readlines()
        for line in lines:
            print(line.strip())
except FileNotFoundError:
    print("文件未找到！")
```

+ ### <font color="purple">**with语句**</font>是一种很实用的语法结构，其主要作用是管理资源，像文件操作、网络连接、数据库连接这类资源，在使用完毕后都<font color="purple">**需要正确释放**</font>，with语句就能很好地处理资源的获取和释放，防止资源泄漏。
+ ### 在执行with语句时，首先获取资源，当with语句块执行完毕，或者在执行过程中出现异常时，会<font color="purple">**自动释放资源**</font>。对于文件操作来说，一般情况下操作完毕，需要调用close方法关闭文件释放资源。但是使用with语句可以简化这一过程。With语句块结束时文件已经自动关闭，无需手动调用`file.close()`

```python
# 写入一个文本文件：
# 写入单行内容
try:
    with open('example.txt', 'w', encoding='utf-8') as file:
        file.write("这是一行新的内容。\n")
except Exception as e:
    print(f"写入文件时出错：{e}")

# 写入多行内容
lines = ["第一行内容。", "第二行内容。", "第三行内容。"]
try:
    with open('example.txt', 'a', encoding='utf-8') as file:
        for line in lines:
            file.write(line + '\n')
except Exception as e:
    print(f"写入文件时出错：{e}")
```

## <font color = #0288d1>处理图像文件</font>
### Python的PIL（Python Imaging Library）库，也就是现在的<font color="purple">**Pillow库**</font>，是处理图像的强大工具。下面介绍如何使用 Pillow库打开、读取、显示图像文件，以及将图像文件转换为 NumPy数组。在开始之前，<font color="purple">**需要安装Pillow和NumPy库**</font>。

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 1. 打开图像文件
try:
    # 请将 'example.jpg' 替换为你实际的图像文件路径
    image = Image.open('example.jpg')
    print("图像已成功打开。")

    # 2. 显示图像（使用 matplotlib）
    plt.imshow(image)
    plt.axis('off')  # 关闭坐标轴
    plt.show()

    # 3. 将图像转换为 NumPy 数组
    image_array = np.array(image)
    print(f"图像数组的形状: {image_array.shape}")
    print(f"图像数组的数据类型: {image_array.dtype}")

except FileNotFoundError:
    print("未找到指定的图像文件，请检查文件路径。")
except Exception as e:
    print(f"发生了未知错误: {e}")
```
