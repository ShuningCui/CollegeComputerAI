```python
def is_triangle(x,y,z):
    return x+y>z and x+z>y and y+z>x
```

```python
def triangle_area(x,y,z):
    s=(x+y+z)/2
    return (s*(s-x)*(s-y)*(s-z))**0.5
```

```python
if is_triangle(a1,b1,c1):
    print(triangle_area(a1,b1,c1))
```

```python
if is_triangle(a2,b2,c2):
    print(triangle_area(a2,b2,c2))
```

```python
class Triangle:
    def __init__(self,x,y,z):
        self.x=x
        self.y=y
        self.z=z

    def is_triangle(self):
        return self.x+self.y>self.z and \
        self.x+self.z>self.y and \
        self.y+self.z>self.x

    def triangle_area(self):
        s=(self.x+self.y+self.z)/2
        return (s*(s-self.x)*(s-self.y)*(s-self.z))**0.5
```

```python
t1 = Triangle(a1,b1,c1)
if t1.is_triangle():
    print(t1.triangle_area())

t2 = Triangle(a2,b2,c2)
if t2.is_triangle():
    print(t2.triangle_area())
```
