```python
class Triangle(object):
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c
    
    def isTri(self):
        return self.a+self.b>self.c and self.c+self.b>self.a \
                and self.a+self.c>self.b
    
    def tarea(self):
        s=(self.a+self.b+self.c)/2.0
        return (s*(s-self.a)*(s-self.b)*(s-self.c))**0.5
```

```python
x=3.0
y=4.0
z=5.0

t1=Triangle(x,y,z)
print(t1.tarea())
```
