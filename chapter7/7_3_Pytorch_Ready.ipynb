## <font color = #0288d1>张量（tensor）</font>

+ ### 一个标量或者一个数字是0维的；
+ ### 一个向量是1维的；
+ ### 一个矩阵是2维的，
+ ### 3维或更多维我们称为<font color="purple">**张量**</font>。
+ ### 但是，为了简单起见，将向量和矩阵也可以称为张量。
+ ### 因此，从现在开始，<font color="purple">**一切都是标量或张量**</font>。
+ ### 使用`tensor()`可以创建标量或张量。
+ ### 张量函数，`ones()`、`zeros()`、`rand()`、`randn()`等。

```python
import torch
scalar = torch.tensor(3.14159)
vector = torch.tensor([1, 2, 3])
matrix = torch.ones((2, 3), dtype=torch.float)
tensor_n = torch.randn((2, 3, 4), dtype=torch.float)
tensor = torch.rand((2, 3, 4), dtype=torch.float)
print(scalar)
print(vector)
print(matrix)
print(tensor)
print(tensor_n)
```

### 可以用tensor的`size()`方法或者shape属性得到张量的维度：

```python
print(tensor.size(), tensor.shape)
```

+ ### 还可以使用`view()`方法改变张量的维度。
+ ### 需要注意的是<font color="purple">**view()方法只返回具有所需形状的张量，该张量与原始张量共享基础数据，它不会创建一个新的、独立的张量**</font>：

```python
import torch
matrix = torch.rand(4,6)
same_matrix = matrix.view(2,2,-1)
same_matrix
```

+ ### 如此前创建的matrix张量是一个2\*3的张量，可以重新塑造为一个1*6的张量。但是这两个张量的所有数据是<font color="purple">**共享**</font>的
+ ### 如果需要复制一个新的张量副本，PyTorch推荐使用`clone()`加上`detach()`方法的组合：

```python
another_matrix = matrix.view(1, 6).clone().detach()
```

+ ### PyTorch总是接受张量类型的数据，把numpy数组转化为张量，可以使用`as_tensor()`
+ ### 需要注意的是和view方法一样，`as_tensor()`并不产生新的数据，而是和numpy共用同一个数据。
+ ### 因此如果改变的原来的numpy数组的数据，张量也会随之改变，反之亦然。
+ ### 如果你确实需要一个新的副本，请使用一开始介绍的`torch.tensor()`。
+ ### 最后可以使用`numpy()`方法将张量转变回numpy数组。

```python
import numpy as np
x_train = np.random.rand(100, 10).astype(np.float32)
x_train_tensor = torch.as_tensor(x_train)
```

## <font color = #0288d1>一个线性模型在神经网络中可以看作是一个层</font>

![](../pictures/7-12.png)

+ ### 需要做的是<font color="purple">**从Module类派生出一个自己的类**</font>，来组合创建需要的模型。
+ ### 在这个派生类中，必须提供2个函数。一个是初始化函数`__init__()`，该函数就是类的构造函数用于初始化参数。
+ ### 还有一个是`forward()`函数，PyTorch调用该函数计算模型的预测值。

```python
import torch.nn as nn

class LinearExample(nn.Module):
    def __init__(self):
        """初始化线性层"""
        super().__init__()
        self.linear01 = nn.Linear(3,5,bias=True)
        """定义线性层，输入维度为3，输出维度为5"""
        self.linear02 = nn.Linear(5, 1)

    def forward(self, x):
        """前向传播"""
        x = self.linear01(x)
        y = self.linear02(x)
        return y
```

```python
model = LinearExample()
```
